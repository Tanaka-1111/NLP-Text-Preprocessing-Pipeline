# --- CSVから読み込み（確認や後続処理用） ---

```python
fname = '/content/drive/MyDrive/Colab Notebooks/files/parsed_imdb_train_X.csv'
X_train = pd.read_csv(fname,header=None)
X_train
```

```python
fname = '/content/drive/MyDrive/Colab Notebooks/files/parsed_imdb_train_y.csv'
y_train = pd.read_csv(fname, header=None)
y_train
```

```python
fname = '/content/drive/MyDrive/Colab Notebooks/files/parsed_imdb_test_X.csv'
X_test = pd.read_csv(fname,header=None)
X_test
```

```python
fname = '/content/drive/MyDrive/Colab Notebooks/files/parsed_imdb_test_y.csv'
y_test = pd.read_csv(fname,header=None)
y_test
```

# --- ラベルのOne-hotエンコーディング ---

```python
from keras.utils import to_categorical
y_train_onehot = to_categorical(y_train)
```

```python
print("X_train:", X_train.shape)
print("y_train_onehot:", y_train_onehot.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

## **②通常のニューラルネットワークで機械学習するプログラムを作成**


```python
from keras.models import Sequential
from keras.layers import Dense

model = Sequential([
  Dense(input_dim=100, units=64, activation='relu'),
  Dense(input_dim=64, units=32, activation='relu'),
  Dense(units=2, activation='softmax')
])
```

```python
model.compile(
    loss='categorical_crossentropy',
    optimizer='adam'
  )

model.summary()
```

```python
model.fit(
  X_train, y_train_onehot,
  epochs=30
)
```

```python
import numpy as np

prob = model.predict(X_test)
y_pred = np.argmax(prob,axis=1)
y_pred
```

```python
from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred) #precision, recall
print(report)
```

```python
from sklearn.metrics import confusion_matrix

c_matrix = confusion_matrix(y_test, y_pred)
print(c_matrix)
```

# **③RNNで機械学習するプログラムを作成**

```python
from keras.layers import Embedding, SimpleRNN

dim = X_train.max().max() + 1

model_rnn = Sequential([
  Embedding(dim, 32),
  SimpleRNN(16),
  Dense(2, activation='softmax')
])
```

```python
model_rnn.compile(
  loss = 'binary_crossentropy',
  optimizer = 'adam'
)

model_rnn.summary()
```

```python
model_rnn.fit(
  X_train, y_train_onehot,
  epochs=10
)
```

```python
prob = model_rnn.predict(X_test)
y_pred = np.argmax(prob, axis=1)
y_pred
```

```python
report = classification_report(y_test, y_pred) #precision, recall
print(report)
```

```python
c_matrix = confusion_matrix(y_test, y_pred) #混合行列
print(c_matrix)
```
